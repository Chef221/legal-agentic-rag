```python
from hashlib import sha256
from pathlib import Path
import runpy
import sys
from zipfile import ZipFile

input_root = Path('/kaggle/input')
working = Path('/kaggle/working')
scripts = sorted(input_root.rglob('m48_kaggle_candidate_dev.py'))
if not scripts:
    zips = sorted(input_root.rglob('legal-agentic-rag-m48-source.zip'))
    assert len(zips) == 1, f'Cần đúng 1 source ZIP M48, tìm thấy: {zips}'
    bootstrap = working / 'm48-dev-bootstrap'
    if not bootstrap.is_dir():
        with ZipFile(zips[0]) as archive:
            archive.extractall(bootstrap)
    scripts = sorted(bootstrap.rglob('m48_kaggle_candidate_dev.py'))
assert scripts, 'Không tìm thấy dev runner M48'
digests = {sha256(path.read_bytes()).hexdigest() for path in scripts}
assert len(digests) == 1, f'Có nhiều source M48 khác nhau: {scripts}'
script = scripts[0]
sys.path.insert(0, str(script.parent))
runpy.run_path(str(script), run_name='__main__')
```
