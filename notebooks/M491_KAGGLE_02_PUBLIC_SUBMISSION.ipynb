Chỉ chạy sau khi dev promotion. Add: source M49.1; `public-official.json`; output M45; output training M49. Bật Internet và GPU T4 x2. Batch checkpoint sau từng câu.

```python
from hashlib import sha256
from pathlib import Path
import runpy
import sys
from zipfile import ZipFile

root = Path('/kaggle/input')
working = Path('/kaggle/working')
scripts = sorted(root.rglob('m491_kaggle_public_submission.py'))
if not scripts:
    zips = sorted(root.rglob('legal-agentic-rag-m491-source.zip'))
    assert len(zips) == 1, f'Cần đúng 1 source ZIP M49.1: {zips}'
    target = working / 'm491-public-bootstrap'
    if not target.is_dir():
        with ZipFile(zips[0]) as archive:
            archive.extractall(target)
    scripts = sorted(target.rglob('m491_kaggle_public_submission.py'))
assert scripts, 'Không tìm thấy public runner M49.1'
digests = {sha256(path.read_bytes()).hexdigest() for path in scripts}
assert len(digests) == 1, f'Có nhiều source M49.1 khác nhau: {scripts}'
script = scripts[0]
sys.path.insert(0, str(script.parent))
runpy.run_path(str(script), run_name='__main__')
```
